## Phase 1: Data Preprocessing

### Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', '{:.2f}'.format)

print('Libraries imported successfully!')
```

### Load Dataset

```python
df = pd.read_csv('D:/Office/group task/AI-Powered-Credit-Risk-Loan-Default-Prediction-System/loan_default_project/data/Loan_default.csv')

print(f'Dataset loaded successfully!')
print(f'Total Rows    : {df.shape[0]:,}')
print(f'Total Columns : {df.shape[1]}')
```

### Understand Columns

```python
print("--- Dataset Information ---")
df.info()

print("\n--- Statistical Summary ---")
df.describe()
```

### Remove IDs

```python
df = df.drop(columns=['LoanID'])

print("Remaining Columns in the Dataset:")
print(df.columns.tolist())
```

### Handle Missing Values

```python
missing_values = df.isnull().sum()
print("Missing Values per Column:")
print(missing_values)
```

### Remove Duplicates

```python
duplicate_count = df.duplicated().sum()
print(f"Number of duplicate rows found: {duplicate_count}")

if duplicate_count > 0:
    df = df.drop_duplicates()
    print("Duplicates successfully removed.")
```

### Fix Data Types & Standardize Categories

```python
categorical_cols = df.select_dtypes(include=['object']).columns

for col in categorical_cols:
    df[col] = df[col].astype(str).str.strip()

print("Data types checked and categories standardized.")
```

### Encode Categories

```python
binary_cols = ['HasMortgage', 'HasDependents', 'HasCoSigner']
for col in binary_cols:
    df[col] = df[col].map({'Yes': 1, 'No': 0})

multiclass_cols = ['Education', 'EmploymentType', 'MaritalStatus', 'LoanPurpose']
df = pd.get_dummies(df, columns=multiclass_cols, drop_first=True)


df.head()
```

### Check Outliers

```python
numerical_cols = ['Age', 'Income', 'LoanAmount', 'CreditScore', 'MonthsEmployed', 'NumCreditLines', 'InterestRate', 'LoanTerm', 'DTIRatio']

print("--- Outlier Detection using IQR ---")
for col in numerical_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
   
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    print(f"{col}: Found {len(outliers)} outliers.")
```

### Final Review

```python

print(f"Final Dataset Shape: {df.shape}")

df.info()
```

### Save Clean Dataset

```python
df.to_csv('Cleaned_Loan_Default.csv', index=False)

print("Preprocessing complete! The file has been saved as 'Cleaned_Loan_Default.csv'.")
```
